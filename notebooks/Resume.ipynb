Para un proyecto de **regresión** (donde tu objetivo es predecir un valor numérico continuo, como el precio de una casa en tu dataset de *House Prices*), existe una gran variedad de algoritmos que van desde los más sencillos e interpretables hasta modelos avanzados de alto rendimiento.

Aquí tienes los principales modelos de regresión agrupados por su complejidad y tipo de funcionamiento:

---

### 1. Modelos Lineales

Son los más sencillos, rápidos de entrenar y altamente interpretables. Funcionan buscando una relación lineal (una línea recta o un plano) entre tus variables de entrada y el precio objetivo.

* **Regresión Lineal Simple / Múltiple:** El modelo clásico. Estima los coeficientes (pesos) minimizando el error cuadrático medio. Es ideal para empezar y establecer una **línea base (*baseline*)**.
* **Regresión Ridge ($L2$):** Una variante de la regresión lineal que añade una penalización a los pesos grandes para evitar el sobreajuste (*overfitting*). Muy útil si tienes muchas variables o multicolinealidad.
* **Regresión Lasso ($L1$):** Similar a Ridge, pero con una penalización diferente que puede reducir los coeficientes de las variables menos importantes exactamente a **cero**, actuando también como un selector automático de características.
* **Regresión ElasticNet:** Combina lo mejor de Ridge y Lasso (penalizaciones $L1$ y $L2$ simultáneamente).

---

### 2. Modelos Basados en Árboles de Decisión

En lugar de trazar líneas rectas, estos modelos dividen los datos en ramas formando una estructura de árbol basada en preguntas (por ejemplo: *¿El área del sótano es mayor a 100 m²?*). Son excelentes capturando relaciones no lineales.

* **Decision Tree Regressor (Árbol de Decisión):** Un solo árbol de decisión. Es muy fácil de visualizar, pero tiende a sobreajustar fácilmente si se le permite crecer sin límite.
* **Random Forest Regressor (Bosque Aleatorio):** Combina cientos de árboles de decisión entrenados en paralelo con subconjuntos aleatorios de datos (*Bagging*). Es uno de los modelos más potentes, estables y resistentes al sobreajuste por defecto.
* **Gradient Boosting Regressor:** Entrena árboles de manera secuencial, donde cada árbol nuevo intenta corregir los errores (residuos) cometidos por el árbol anterior (*Boosting*).
* **XGBoost, LightGBM y CatBoost:** Implementaciones ultraoptimizadas y avanzadas de Gradient Boosting. Son los reyes indiscutibles en competencias de ciencia de datos (como Kaggle) cuando se trabaja con datos tabulares estructurados debido a su velocidad y precisión.

---

### 3. Modelos de Vecinos Cercanos y de Distancia

* **K-Nearest Neighbors Regressor (KNN):** Predice el valor de una casa basándose en el promedio de los $K$ vecinos más cercanos en el espacio de características. Es simple, pero puede volverse lento con datasets muy grandes.
* **Support Vector Regressor (SVR):** Utiliza vectores de soporte para encontrar una función que se ajuste a los datos dentro de un margen de tolerancia aceptable. Funciona muy bien en espacios de alta dimensionalidad.

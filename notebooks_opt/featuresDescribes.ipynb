```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
%matplotlib inline
```

```python

PATH_DATASET = "../data/raw/train.csv"
df = pd.read_csv(PATH_DATASET)
print("dimensiones del dataset:", df.shape)
display(df.head())
```

```python
display(df.describe())
```

```python
missingValues = df.copy().drop('Id', axis=1)
rows, columns = missingValues.shape
missingValues = pd.DataFrame(missingValues.isnull().sum())
missingValues = missingValues.loc[missingValues[0] > 0]
missingValues['porcentaje'] = (missingValues.iloc[:, 0] / rows * 100).round(2)
missingValues.rename(columns={0: 'faltantes'}, inplace=True)
missingValues = missingValues.sort_values(by='faltantes', ascending=False)
missingValues
```
